```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap, Normalize
from mpl_toolkits.mplot3d import Axes3D
import os
from sklearn.decomposition import PCA
import matplotlib as mpl

mpl.rcParams['font.family'] = 'Helvetica'
mpl.rcParams['font.sans-serif'] = ['Helvetica']  # ensures fallback


colors_list = ["#161347","#304c99", "#62a1db", "#e7d87d", "#dd9f40", "#b4451f", "#b01111","#7a0808","#4f0000"]
cmap = LinearSegmentedColormap.from_list("reinterpret_map", colors_list, N=256)
```

```python
current_directory = os.getcwd()

file_path = os.path.join(
    current_directory,
    "../../data/subject_by_node_Viewing1_Viewing2_subs_schaefer_rois.csv"
)

data = pd.read_csv(file_path)
```

```python
# Keep only experimental condition
data = data[data["Condition"] == "expt"]

# Rename to match R logic
data["appraisal_self"]  = data["appraisal_SELF_cond1"]
data["appraisal_other"] = data["appraisal_OTHER_cond1"]

# Semantic distance = 1 - cosine similarity
data["semantic_distance"] = 1 - data["cosine_sim_avg_inputs_BERT_context_miniLM_L6"]

# Only node 41
data_full = data[data["Node_Number"] == 41].copy()
```

```python
data_full["confidence_scaled"]      = data_full["confidence"] / 100
data_full["appraisal_self_scaled"]  = data_full["appraisal_SELF_cond1"] / 100
data_full["appraisal_other_scaled"] = data_full["appraisal_OTHER_cond1"] / 100

data_full["reinterpretation"] = (
    data_full["confidence_scaled"]
    - data_full["appraisal_self_scaled"]
    + data_full["appraisal_other_scaled"]
)
```

```python
# Fixed normalization range
norm = Normalize(vmin=-1, vmax=2)
y = data_full["appraisal_self_scaled"].values
z = data_full["confidence_scaled"].values
x = data_full["appraisal_other_scaled"].values
c = data_full["reinterpretation"].values

fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, projection='3d')
# Make 3D background white instead of gray
ax.w_xaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))
ax.w_yaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))
ax.w_zaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))

# Also remove the darker gridlines on panes if desired
ax.w_xaxis.line.set_color("black")
ax.w_yaxis.line.set_color("black")
ax.w_zaxis.line.set_color("black")
ax.w_xaxis.line.set_linewidth(2)
ax.w_yaxis.line.set_linewidth(2)
ax.w_zaxis.line.set_linewidth(2)
#ax.view_init(elev=20, azim=15)
ax.view_init(elev=25, azim=20)
ax.set_xlim(1, 0)



sc = ax.scatter(
    x, y, z,
    c=c,
    cmap=cmap,
    norm=norm,
    s=80,
    edgecolor='k',
    linewidth=0.5,
    alpha = 1,
)
# Draw vertical projection lines
for xi, yi, zi, ci in zip(x, y, z, c):
    ax.plot(
        [xi, xi],        # x goes from xi → xi (vertical)
        [yi, yi],        # y goes from yi → yi
        [0, zi],         # z from bottom plane → point height
        color=cmap(norm(ci)),
        linewidth=1.2,
        alpha=1
    )
ax.tick_params(axis='both', which='major', labelsize=14)



ax.set_xlabel("other")
ax.set_ylabel("self2")
ax.set_zlabel("self1")

cbar = plt.colorbar(sc, pad=0.1,    orientation="vertical", shrink=.6)

cbar.set_label("Reinterpretation (–1 to 2)", fontsize=14)



plt.tight_layout()
plt.savefig("../../results/reinterpretation_plots/reinterpretation_3D_scatter.pdf",transparent=True)
plt.show()
```

# Trying som kind of manifold thing

```python

# Select behavioral variables
X = data_full[
    ["appraisal_self_scaled",
     "confidence_scaled",
     "appraisal_other_scaled"]
]

# remove inf values
X = X.replace([np.inf, -np.inf], np.nan)

# remove missing rows
X = X.dropna()

# keep matching rows
data_clean = data_full.loc[X.index]

# convert to numpy
X = X.values

pca = PCA()

embedding = pca.fit_transform(X)

print("Variance explained:", pca.explained_variance_ratio_)

pc1 = embedding[:,0]
pc2 = embedding[:,1]
pc3 = embedding[:,2]

loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1","PC2","PC3"],
    index=["Self","Confidence","Other"]
)

print(loadings)
```

```python
corr_pc2 = np.corrcoef(pc2, data_clean["reinterpretation"])[0,1]

if corr_pc2 < 0:
    pc2 = -pc2
    loadings["PC2"] = -loadings["PC2"]
    
pc1 = -pc1
loadings["PC1"] = -loadings["PC1"] #flipping one too so that higher magnitude is higher
    
print(np.corrcoef(pc1, data_clean["reinterpretation"])[0,1])

print(np.corrcoef(pc2, data_clean["reinterpretation"])[0,1])
```

```python
loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1","PC2","PC3"],
    index=["Self","Confidence","Other"]
)

print(loadings)
#PC1 -- Because Self, Confidence, and Other all load in the same direction, PC1 reflects trials where participants give globally higher vs. lower ratings across all measures.
#PC1 captures overall appraisal magnitude, reflecting trials in which participants gave stronger versus weaker ratings across self-appraisal, confidence, and other-appraisal.
#Can call it overall response magnitude, global appraisal strength, general evaluative intensity

#PC2 -- PC2 is dominated by the large negative loading for Other and positive loadings for Self and Confidence.
#PC2 captures a perspective axis contrasting other-appraisal with self-appraisal and confidence.

#Principal component analysis revealed that behavioral responses lay on a low-dimensional manifold: the first two components explained 91% of the variance in appraisal space. The first component (PC1) reflected overall appraisal magnitude, with positive loadings from self-appraisal, confidence, and other-appraisal. The second component (PC2) captured a perspective contrast between self-oriented and other-oriented interpretations.
    
print(np.corrcoef(pc1, data_clean["reinterpretation"])[0,1])
print(np.corrcoef(pc2, data_clean["reinterpretation"])[0,1])
#reinterpretation 
# So reinterpretation is not mainly about stronger or weaker responses (PC1).
#Instead it reflects shifts in interpretational perspective, good!! 
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))

plt.scatter(
    pc1,
    pc2,
    c=data_clean["reinterpretation"],
    cmap=cmap,
    norm=norm,
    s=60,
    edgecolor="k",
    alpha=0.7
)

plt.xlabel("PC1 (response strength, low to high)")
plt.ylabel("PC2 (Other–Self interpretation)")

cbar = plt.colorbar()
cbar.set_label("Reinterpretation")

plt.gca().set_aspect("equal")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# -------------------------------------------------------
# 1. Prepare data
# -------------------------------------------------------
df = data_full.copy()

df["semantic"]   = df["semantic_distance"]
df["confidence"] = df["confidence_scaled"]
df["other_minus_self"] = df["appraisal_other_scaled"] - df["appraisal_self_scaled"]

# -------------------------------------------------------
# 2. Fit model (NO mean-centering)
# -------------------------------------------------------
model = smf.ols(
    "other_minus_self ~ semantic * confidence",
    data=df
).fit()

print(model.summary())

# -------------------------------------------------------
# 3. Create prediction grid (AXES SWITCHED)
#    X-axis = confidence
#    Y-axis = semantic distance
# -------------------------------------------------------
confidence_range = np.linspace(df["confidence"].min(), df["confidence"].max(), 200)
semantic_range   = np.linspace(df["semantic"].min(), df["semantic"].max(), 200)

grid = pd.DataFrame(
    [(c, s) for c in confidence_range for s in semantic_range],
    columns=["confidence", "semantic"]
)

grid["pred"] = model.predict(grid)
Z = grid["pred"].values.reshape(len(confidence_range), len(semantic_range))

# -------------------------------------------------------
# 4. Build custom colormap
# -------------------------------------------------------
colors = [
    (0.0, "#4c2051"),  # low end (purple)
    (0.25, "#87378e"),
    (0.5, "white"),    # midpoint (0)
    (0.75, "#4a784f"),
    (1.0, "#2a442c"),  # high end (green)
]

cmap = LinearSegmentedColormap.from_list("custom_diverging", colors)

# -------------------------------------------------------
# 5. Plot heatmap + zero contour
# -------------------------------------------------------
plt.figure(figsize=(8, 6))

vmax = 0.65
vmin = -vmax
levels = np.linspace(vmin, vmax, 200)

cont = plt.contourf(
    confidence_range,
    semantic_range,
    Z.T,
    levels=levels,
    cmap=cmap
)

# Colorbar
# cbar = plt.colorbar(cont)
# cbar.set_label("OTHER − SELF")
# cbar.set_ticks([vmin, 0, vmax])
# cbar.set_ticklabels(["−0.65", "0", "0.65"])

# Zero contour line
zero_contour = plt.contour(
    confidence_range,
    semantic_range,
    Z.T,
    levels=[0],
    colors="black",
    linewidths=2,
    antialiased=True
)

plt.clabel(zero_contour, inline=True, fontsize=12, fmt="Other = Self")

plt.gca().set_aspect("equal", adjustable="box")

# Labels & title
plt.xlabel("Confidence (Self₁)")
plt.ylabel("Semantic distance")
# plt.title("Predicted OTHER − SELF\nConfidence × Semantic distance")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["semantic"],
    df["confidence"],
    c=df["other_minus_self"],
    cmap=cmap,
    vmin=-0.65,
    vmax=0.65,
    s=40,
    edgecolor="none",
    alpha=0.9
)

cbar = plt.colorbar()
cbar.set_label("OTHER − SELF")
cbar.set_ticks([-0.65, 0, 0.65])

plt.xlabel("Semantic distance")
plt.ylabel("Confidence (Self₁)")

plt.tight_layout()
plt.show()
```
